## **Interpretation**
Initial checks showed that neither review_id nor order_id is unique.


*   Duplicate review_id records were inspected using:
order_reviews[order_reviews.duplicated(subset='review_id', keep=False)]

These records shared identical review content, score, and timestamps, while being associated with different order_id values.

*   Duplicate order_id records were inspected using:
order_reviews[order_reviews.duplicated(subset='order_id', keep=False)]

These records contained different review_id values and often different review content and timestamps.

This is interesting. In the first case, it seems like a single review may be linked to multiple orders. In the second case, one order receives multiple reviews. This is more familiar, because someone can change the review after buying one product. For example, at the first glance, one gives a five star review based on the delivery time, but then after using the product, doesnt feel satisfied and changes the review to a 3 star review. One thing is for sure. Every row of this table is an association between an order and a review record.

## **Conclusion**
The repeated IDs were not treated as duplicate rows because the records contain meaningful relationship information and different combinations of order_id and review_id. No rows were removed during this step

The absence of the values could occur to the fact that not all deliveries were delivered.

```python
orders.groupby("order_status")[
    "order_delivered_customer_date"
].agg(
    total="size",
    missing=lambda x: x.isna().sum()
)
```

This hypothesis seems correct according to the above table. For the order_delivered_customer_date column, the most missing values come from orders that have cancelled, shipped or unavailable on their status, which make total sense. The strange thing here is that 6 orders that were cancelled seem to have a delivery date, which is interesting to look further into.
```python
orders[
    (orders["order_status"] == "delivered") &
    (orders["order_delivered_customer_date"].isna())
]
```

Among the 96,478 orders with status "delivered", only 8 records are missing the customer delivery timestamp. Most of these orders still contain valid approval and carrier timestamps, suggesting that the orders were likely completed but the final delivery timestamp was not recorded. One record is missing both carrier and customer delivery timestamps despite being marked as delivered, indicating a potential data quality issue. Because the inconsistent values are too small of a quantity compared to the rest of the data, the values will stay NaN. If further analysis occurs, it will be considered.
